```python
# criar uma coluna booleana

# acessando a coluna "frequencia de aposta"
dados.freq_numerica
```

```python
dados["dados"] = dados.qual_é_o_seu_nível_de_escolaridade_?
dados
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Calculate the mean 'freq_numerica' for each schooling level
escolaridade_aposta_media = dados.groupby('Qual é o seu nível de escolaridade?')['freq_numerica'].mean().sort_values(ascending=False)

# Display the results
display(escolaridade_aposta_media)

# Plot the results
plt.figure(figsize=(12, 7))
sns.barplot(x=escolaridade_aposta_media.index, y=escolaridade_aposta_media.values, palette='viridis')
plt.title('Média de Frequência de Apostas por Nível de Escolaridade')
plt.xlabel('Nível de Escolaridade')
plt.ylabel('Média de Frequência de Apostas')
plt.xticks(rotation=45, ha='right') # Rotate labels for better readability
plt.tight_layout() # Adjust layout to prevent labels from overlapping
plt.show()
```
